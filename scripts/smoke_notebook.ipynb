```python
import warnings
import igraph as ig
from tau_community_detection import run_clustering
```

```python
import loky
print(f'loky {loky.__version__} available')
```

```python
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter('always')
    g = ig.Graph.Famous('Zachary')
    clustering = run_clustering(g, random_seed=42, population_size=20, max_generations=5)

loky_warnings = [w for w in caught if 'loky' in str(w.message).lower()]
assert not loky_warnings, f'Unexpected loky warning: {loky_warnings[0].message}'
assert hasattr(clustering, 'modularity') and isinstance(clustering.modularity, float)
assert len(clustering) >= 1
print(f'PASS  no loky warning emitted')
print(f'PASS  {len(clustering)} communities, modularity={clustering.modularity:.4f}')
```
